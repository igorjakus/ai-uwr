## Podstawowe opcje

| --------- | --------------------------------- | -------------------- |
| `label`   | nazwa/identyfikator komórki       | `#\| label: wykres1` |
| `eval`    | czy kod ma być wykonany           | `#\| eval: true`     |

| `error`   | czy pokazać błędy w wyniku        | `#\| error: true`    |

## Opcje wykresów

| Opcja        | Opis                           | Przykład                   |
| ------------ | ------------------------------ | -------------------------- |
| `fig-width`  | szerokość wykresu              | `#\| fig-width: 5`         |
| `fig-height` | wysokość wykresu               | `#\| fig-height: 4`        |
| `fig-asp`    | proporcje wykresu              | `#\| fig-asp: 1.5`         |
| `fig-cap`    | podpis wykresu                 | `#\| fig-cap: "Histogram"` |
| `fig-align`  | wyrównanie                     | `#\| fig-align: center`    |
| `fig-pos`    | pozycja wykresu w PDF/LaTeX    | `#\| fig-pos: H`           |
| `dpi`        | rozdzielczość                  | `#\| dpi: 300`             |
| `out-width`  | szerokość wykresu w dokumencie | `#\| out-width: 80%`       |
| `out-height` | wysokość wykresu w dokumencie  | `#\| out-height: 400px`    |



### Przykład wykresu



```r
#| label: fig-histogram
#| fig-cap: "Histogram zmiennej X"
#| fig-width: 10
#| fig-height: 20
#| fig-align: center
#| dpi: 150

X = np.random.randn(100)

plt.hist(X, density=True)
plt.show()
```
